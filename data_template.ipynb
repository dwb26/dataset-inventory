# From Data to Decisions: Building a Roadmap for a Synthetic Population for Public Health Policy Analysis

## Executive Summary

High-level overview of the 12-month seed funding project, outlining the integration of the four PHI UK themes to model clustered exposures.

---

### 1. Introduction & Context

#### 1.1 Project Background

Traditional public health models operate in thematic silos. This project establishes a framework to bridge these boundaries, enabling place-based local policy analysis capable of evaluating clustered exposures (e.g., economic policy intersecting with commercial and built environments).

#### 1.2 Scope of the Four PHI UK Themes

* **Theme 1: Policy Modelling for Health (PMFH):** Focuses on micro-simulating the health equity impacts of national/local economic, tax, and welfare policies.
* **Theme 2: Local Health and Global Profits (LHGP):** Focuses on identifying and regulating the commercial determinants of health at a local scale.
* **Theme 3: Population Mental Health (PMH):** Dedicated to tackling public mental health challenges across lifecourses, tracking rising trends and inequalities.
* **Theme 4: Healthy Urban Places (HUP):** Focuses on leveraging the built, natural, social, and economic environment of places as structural tools for population health.

---

#### 2.2 Methodological Divergences and Commonalities

* **Divergences in Maturity:** Themes 1 and 2 possess operational synthetic population pipelines. Theme 4 is developing trajectories, while Theme 3 relies entirely on empirical datasets.
* **Divergences in Scale:** Geospatial identifiers range from macro-level regional units (OAs/LSOAs) to hyper-granular residential points (UPRNs).
* **Commonalities:** Themes 1 and 2 share historical roots in simulated annealing and rely on Understanding Society as a seed dataset, providing an immediate opportunity for data harmonisation.
